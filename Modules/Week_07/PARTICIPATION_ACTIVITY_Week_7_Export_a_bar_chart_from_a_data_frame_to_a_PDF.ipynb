```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd

sales = pd.read_csv(
    'InternetSales.csv',
    encoding="latin1",
    index_col= 'OrderDate',
    parse_dates=True)
```

```python
yearly_sales = sales.assign(year=lambda x: x.index.year).groupby("year")["SalesAmount"].sum()
yearly_sales
```

```python
yearly_sales.plot(
    kind="bar",
    color="blue",
    title="Yearly Internet Sales"
)

plt.xlabel("Year")
plt.ylabel("Sales Amount($)")
plt.tight_layout()
plt.savefig("yearly_sales.pdf", format="pdf")
```
