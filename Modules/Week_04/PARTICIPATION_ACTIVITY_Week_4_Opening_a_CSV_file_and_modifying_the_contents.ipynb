```python
import pandas as pd
import os
os.chdir('/content/drive/MyDrive/MSSP 607/WeeklyModules/Week_4')

df = pd.read_csv('parsed.csv')
df.head()
print(df[df['parsed_place'] == 'Antarctic'])
```

```python
# 2a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan_mb_95th = df[
    (df['parsed_place'] == 'Japan') &
     (df['magType'] == 'mb')]['mag'].quantile(0.95)
print(f'The 95th percentile of earthquake magnitude in Japan is {japan_mb_95th}.')
```

```python
# 2b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
indonesia_tsunami = df[(df['parsed_place'] == 'Indonesia') & (df['tsunami'] == 1)]
percentage_eq_ts = (indonesia_tsunami.shape[0])  / ((df[df['parsed_place'] == 'Indonesia']).shape[0])
print(f"The percentage of earthquakes in Indonesia that were coupled with tsunamis is {percentage_eq_ts * 100:.2f}%.")
```

```python
# 2c. Get summary statistics for earthquakes in Nevada
df[df['parsed_place'] == 'Nevada'].describe()
```

```python
# 2d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.
ring_of_fire_text = """Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico, Japan, Philippines, Indonesia, New Zealand, Antarctic, Canada, Fiji, Alaska, Washington, California, Russia, Taiwan, Tonga, Kermadec Islands"""
ring_of_fire = ring_of_fire_text.split(', ')
print(ring_of_fire)
df['ring_of_fire'] = df['parsed_place'].isin(ring_of_fire)
df.loc[df['parsed_place'].str.contains('Antarctic|Kermadec', na=False),
       "ring_of_fire"] = True
print(df[df['ring_of_fire']]['parsed_place'].unique()) #check if all countries are included
```

```python
# 2e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
rof_earthquakes = df['ring_of_fire'].value_counts()
print(rof_earthquakes)
```

```python
# 2f. Find the tsunami count along the Ring of Fire.
rof_tsnamis = df[(df['ring_of_fire'] == True) & (df['tsunami'] == 1)]
print(f"The tsunami count along the Ring of Fire is {rof_tsnamis.shape[0]}.")
```
