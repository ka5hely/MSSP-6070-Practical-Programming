```python
letters = ['a', 'b', 'c', 'd']

result_3b = letters[int(int('3'*2)//11)]
print(result_3b)
# 3b. evaluates to 'd'
# '3' * 2: replicates the string '3' twice, turning it into '33'.
# int('33'): turns '33' from a string to integer 33.
# int(33//11): integer division results in 3
# letters[3]: access the 4th element in letters, which is 'd'

result_3c = letters[-1]
print(result_3c)
# 3c. evaluates to 'd'
# index -1, being negative, will access the first element from the end of the list
# which is the last element 'd'

result_3d = letters[:2]
print(result_3d)
#3d. results in ['a', 'b']
# [:2] means taking elements from index 0 to but not including index 2
# In letters list, 'a' is at index 0, 'b' is at index 1
```

```python
spam = ['apple', 'bananas', 'tofu', 'cats']

def comma_code_prac(elements):
  if len(elements) == 0:
    return 'Error: list is empty'

  if len(elements) == 1:
    return elements[0]

  result = ''
  for i in range(len(elements) - 1):
    result = result + elements[i] + ', '
  result += 'and ' + elements[-1]
  return result

print(comma_code_prac(spam))
```
