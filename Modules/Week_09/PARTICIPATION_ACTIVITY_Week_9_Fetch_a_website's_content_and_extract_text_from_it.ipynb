```python
#!pip -q install beautifulsoup4
#!pip -q install requests

import os, time, requests, bs4
```

```python
# I chose The Met Collection, which has an Open Access section
# The URL is obtained from https://metmuseum.github.io/
# https://collectionapi.metmuseum.org/public/collection/v1.1/search

photo_category = input("Enter photo category (eg. dog): ")
```

```python
url = "https://collectionapi.metmuseum.org/public/collection/v1.1/search?q=" + photo_category + '&hasImages=true&limit=10'

# As a simple exercise, I'll only take 10 photos
print("Searching...")
print("Page URL:", url)
```

```python
# Download the web page
res = requests.get(url)
res.raise_for_status()

# Convert the API's JSON response into Python data
data = res.json()
data
```

```python
# Create a folder in the working directory for the images
folder_name = photo_category + "_images"
os.makedirs(folder_name, exist_ok=True)

# We need the artwork's ID numbers to download
art_ids = data["objectIDs"]

image_num = 1

for art_id in art_ids:
  art_url = "https://collectionapi.metmuseum.org/public/collection/v1/objects/" + str(art_id)

  res = requests.get(art_url)
  res.raise_for_status()
  artwork = res.json()
  image_url = artwork["primaryImage"]

  if artwork["isPublicDomain"] == True and image_url != '':

    print("Downloading:", image_url)

    image_res = requests.get(image_url)
    image_res.raise_for_status()

    image_file = open(os.path.join(
        folder_name, 'image_' + str(image_num) + '.jpg'), 'wb')

    for chunk in image_res.iter_content(100000):
      image_file.write(chunk)

    image_file.close()

    image_num += 1
    time.sleep(1)
print("Download Complete.")
```
