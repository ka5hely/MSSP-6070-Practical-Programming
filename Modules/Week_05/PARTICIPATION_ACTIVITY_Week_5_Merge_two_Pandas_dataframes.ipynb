```python
# 2b. Read each file into a dataframe

import pandas as pd

aapl = pd.read_csv('aapl.csv')
amzn = pd.read_csv('amzn.csv')
fb = pd.read_csv('fb.csv')
goog = pd.read_csv('goog.csv')
nflx = pd.read_csv('nflx.csv')
```

```python
# 2c. Add a column to each dataframe, called 'ticker'

aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
nflx['ticker'] = 'NFLX'
goog['ticker'] = 'GOOG'
```

```python
# 2d. Append each file together into a single dataframe

faang_df = pd.concat([aapl, amzn, fb, nflx, goog], ignore_index=True)
faang_df
```

```python
# 2e. Save the results in a CSV file called 'faanf.csv'
faang_df.to_csv('faang.csv', index=False)
```

```python
# 2f. Use type conversion to change the 'date' column into a 'datetime'
# and the 'volume' column into integers.
# Then sort by 'date' and 'ticker.'

faang = pd.read_csv('faang.csv')

faang = faang.assign(
    date = pd.to_datetime(faang['date']),
    volume = faang['volume'].astype('int')
).sort_values(
    ['date', 'ticker']
)

faang.head()
```

```python
# 2g. Find the seven rows within the highest value for volume.

faang.nlargest(n=7, columns='volume')
```

```python
# 2h. Use the melt() method to make the 'date' and 'ticker' key identifiers.
# Melt the remaining columns so that there are not separated.

faang_melt = faang.melt(
    id_vars=['date', 'ticker'],
    value_vars=['open', 'high', 'low', 'close', 'volume'])
faang_melt
```
