## Exercise 1
We want to look at data for the FAANG stocks (Facebook, Apple, Amazon, Netflix, and Google), but we were given each as a separate CSV file. Make them into a single file and store the dataframe of the FAANG data as `faang`:
1. Read each file in.
2. Add a column to each dataframe indicating the ticker it is for.
3. Append them together into a single dataframe.
4. Save the result to a CSV file.

```python
#In this example, I will iterate over the file directory to read each file into
#You can also read each file seperately then append the files together.

faang = pd.DataFrame()
for ticker in ['fb', 'aapl', 'amzn', 'nflx', 'goog']:
    df = pd.read_csv(f'/content/drive/MyDrive/MSSP607/WeeklyModules/Week05/examples/{ticker}.csv')
    # make the ticker the first column
    df.insert(0, 'ticker', ticker.upper())
    faang = pd.concat([faang,df], ignore_index=True)

faang.to_csv('faang.csv', index=False)
```

## Exercise 4
Right now, the data is somewhere between long and wide format. Use `melt()` to make it completely long format.

See: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.melt.html?highlight=melt#pandas.DataFrame.melt



From ChatGPT: In Pandas, *wide* and *long* are two different data formats used to structure data, and each format is suited to different types of analysis or visualization needs. Here’s an overview of the two formats:

### Wide Format
- In the **wide** format, data is spread out across multiple columns, with each column typically representing a different variable.
- This format is common when you have a small number of categorical variables that serve as the columns and rows.
- Each row represents a unique entity, and the columns represent the different attributes of that entity.
  
#### Example of Wide Format
Imagine you have data on students’ test scores for different subjects:

| Student | Math | Science | English |
|---------|------|---------|---------|
| Alice   | 90   | 85      | 95      |
| Bob     | 80   | 75      | 85      |
| Carol   | 70   | 95      | 90      |

Here, each subject is a separate column, and the rows are the students.

### Long Format
- In the **long** format, the data is "melted" so that each row represents a single observation.
- This format has fewer columns: typically, one column for the entity (e.g., Student), one for the variable type (e.g., Subject), and one for the variable value (e.g., Score).
- This format is preferred for analysis methods and visualizations that require data in a tidy format, where each variable is a column and each observation is a row.

#### Example of Long Format
The same data in a long format would look like this:

| Student | Subject | Score |
|---------|---------|-------|
| Alice   | Math    | 90    |
| Alice   | Science | 85    |
| Alice   | English | 95    |
| Bob     | Math    | 80    |
| Bob     | Science | 75    |
| Bob     | English | 85    |
| Carol   | Math    | 70    |
| Carol   | Science | 95    |
| Carol   | English | 90    |

### When to Use Each Format
- **Wide Format**: Useful for situations where you need quick, descriptive statistics or comparisons between variables. It’s also the default format for many data collection tools.
- **Long Format**: Often required for advanced statistical modeling and visualization tools, like `seaborn` or `ggplot`, that expect data in a tidy format.

In Pandas, you can easily convert between these formats using `pd.melt()` (to go from wide to long) and `pd.pivot()` or `pd.pivot_table()` (to go from long to wide).

```python
melted_faang = faang.melt(
    id_vars=['ticker', 'date'],
    value_vars=['open', 'high', 'low', 'close', 'volume']
)
melted_faang.head()
```
