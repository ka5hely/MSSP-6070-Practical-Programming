```python
#3a. The code in a function executes when the function is called
#    not when the function is first defined.

#3b. There is only one global scope, created when your program begins,
#    and is destroyed when the program terminates. A new local scope
#    gets created whenever a program calls a function, and therefore
#    the number of local scopes corresponds with the number of functions.

#3c. When the function call returns, the local scope gets destroyed,
#    along with the variables assigned in the function, but any global
#    global variables will remain in the global scope and retain their values.

#3d. Errors can be handled with try and except statements.
```

```python
# The Collatz Sequence Project
def collatz(number):
    if number % 2 == 0:
        number = number // 2
    else:
        number = number * 3 + 1
    print(number, end=' ')
    return number

while True:
    try:
        user_num = int(input("Enter a number greater than 0: "))
        if user_num <= 0:
            print("Number must be greater than 0!")
            continue
        break
    except ValueError:
        print("Invalid Input: Type something!")

print(user_num, end=' ')
while user_num != 1:
    user_num = collatz(user_num)
```
