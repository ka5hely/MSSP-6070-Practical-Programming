```python
import pandas as pd

df = pd.read_csv("Student Performance Data.csv")
df.head()
```

```python
# I want write a lambda function which takes the df
# and return the students who didn't take test preparation course
# yet still got an above-average total score

no_prep_above_average = lambda df: df[
    (df['test preparation course'] == 'none') &
    (df['total score'] > df['total score'].mean())
]

print(no_prep_above_average(df))
```
