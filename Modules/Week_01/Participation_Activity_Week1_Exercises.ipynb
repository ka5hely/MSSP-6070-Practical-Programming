## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
from statistics import mean
n = len(salaries)
salaries_mean = sum(salaries) / n
print(f"calculated mean = {salaries_mean}")
print(f"verified mean = {mean(salaries)}")
```

### median

```python
from statistics import median
salaries.sort()
if len(salaries) % 2 != 0:
    salaries_median = salaries[int((len(salaries)-1)/2)]
    print(f"calculated median = {salaries_median}")
elif len(salaries) % 2 == 0:
    salaries_median = ((salaries[int((len(salaries)-1)/2)]) + salaries[int(len(salaries)/2)]) / 2
    print(f"calculated median = {salaries_median}")
print(f"verified median = {median(salaries)}")
```

### mode

```python
from statistics import mode
counts = {}
max_count = 0
statistics_mode = 0

for salary in salaries:
    if salary in counts:
        counts[salary] += 1
    else:
        counts[salary] = 1
    if counts[salary] > max_count:
        max_count = counts[salary]
        salaries_mode = salary
print(f"caculated mode = {salaries_mode}")
print(f"verified mode = {mode(salaries)}")
```

### sample variance
Remember to use Bessel's correction.

```python
from statistics import variance

sum_sqr_differences = 0
for salary in salaries:
    diff = salary - salaries_mean
    sum_sqr_differences += diff ** 2
salaries_variance = sum_sqr_differences / (n - 1)
print(f"calculated sample variance = {salaries_variance}")
print(f"verified sample variance = {variance(salaries)}")
```

### sample standard deviation
Remember to use Bessel's correction.

```python
from statistics import stdev
salaries_std = salaries_variance ** 0.5
print(f"calculated sample standard deviation = {salaries_std}")
print(f"verified sample variance = {stdev(salaries)}")
```

## Exercise 6: Calculating more statistics
### range

```python
salaries_range = max(salaries) - min(salaries)
print(f"range = {salaries_range}")
```

### coefficient of variation

```python
#standard deviation and mean are already calculated above
salaries_cv = salaries_std / salaries_mean
print(f"coefficient of variation = {salaries_cv}")
```

### interquartile range

```python
mid_point = n // 2
q1_nums = salaries[:mid_point]
if n % 2 != 0:
    q3_nums = salaries[mid_point+1:]
else:
    q3_nums = salaries[mid_point:]
half_mid_point = mid_point // 2
q1 = q1_nums[half_mid_point]
q3 = q3_nums[half_mid_point]
iqr = q3 - q1
print(f"iqr = {iqr}")
```

### quartile coefficent of dispersion

```python
q_coef_dispersion = (q3 - q1) / (q3 + q1)
print(f"quartile coefficient of dispersion = {q_coef_dispersion}")
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled = []
min_salary = min(salaries)
max_salary = max(salaries)
for salary in salaries:
    scaled = (salary - min_salary) / salaries_range
    min_max_scaled.append(round(scaled,4))
print(min_max_scaled)
```

### standardizing

```python
standardized = []
for salary in salaries:
    scaled = (salary - salaries_mean) / salaries_std
    standardized.append(round(scaled, 2))
print(standardized)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
from statistics import covariance
mean_scaled = mean(min_max_scaled)
mean_standardized = mean(standardized)

sum_diff = 0
for i in range(n):
    diff_scaled = min_max_scaled[i] - mean_scaled
    diff_standardized = standardized[i] - mean_standardized
    sum_diff += diff_scaled * diff_standardized
cov = sum_diff / (n-1)
print(cov)
print(covariance(min_max_scaled, standardized))
```

### Pearson correlation coefficient ($\rho$)

```python
from statistics import stdev, correlation

std_scaled = stdev(min_max_scaled)
std_standardized = stdev(standardized)
pearson = cov / (std_scaled * std_standardized)
print(pearson)
print(correlation(min_max_scaled, standardized))
```
